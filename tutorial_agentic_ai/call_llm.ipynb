```python
# to call llm we have to load api keys

from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
# when we run this code, it is going to create an variable environment variable called Google API key and it will load that value from .env file
```

```python
llm = ChatGoogleGenerativeAI(model = "gemma-4-31b-it")

response = llm.invoke("How many moons does jupiter have?")
print(response.text)
```

```python
# system instruction: which means when we are building ai application we can control llm to get single line answer through array

response = llm.invoke([
    ["system", "you are helpful assistant that answers in short one line"],
    ["human", "How many moons does jupiter have?"]
])
print(response.text)
```

```python
# we can also supply a temperature parameter, so when we create a llm we can supply additional temperature parameter

# temperature = 0 means o creativity,rigid,fixed answer. if we want any creativity we have to change temperature to higher value eg: 0.8,0.9 oe 1

# so here i'm writing single tagline for my coffee shop


prompt = "Write a single tagline for a quirky coffee shop. only output the tagline, nothing else."

llm = ChatGoogleGenerativeAI(model = "gemini-2.5-flash", temperature = 0)

response = llm.invoke(prompt)
print(response.text)
```

```python
# here i'm giving higher temperature value for creativity



llm = ChatGoogleGenerativeAI(model = "gemini-2.5-flash", temperature = 1)

response = llm.invoke(prompt)
print(response.text)
```

## call LLM using Groq Models

```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(model = "qwen/qwen3-32b", reasoning_format = "parsed")

messages = [
    (
        "system", 
        "you are a helpful assistant that gives a short answer to a question",
    ),
    (
        "human", 
        "how many moons does jupiter have?"
    ),
]
ai_msg = llm.invoke(messages)
print(ai_msg.content)
```
